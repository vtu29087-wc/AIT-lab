```
from kanren import Relation, facts, var, run, conde

parent = Relation()

facts(parent,
      ("Ram", "Jhony"),
      ("Jhony", "sai"),
      ("Jhony", "Hari"))

def grandparent(x, z):
    y = var()
    return conde((parent(x, y), parent(y, z)))

x = var()

result = run(1, x, grandparent(x, "sai"))
print("Grand parent of sai is:", result[0])
```

```

from kanren import Relation, facts, var, run, conde

parent = Relation()

facts(parent,
      ("Ram", "Ravi"),
      ("Ram", "Priya"),
      ("Kumar", "Ravi"),
      ("Kumar", "Priya"))

def sibling(x, y):
    p = var()
    return conde((parent(p, x), parent(p, y)))

x = var()
result = run(0, x, sibling(x, "Ravi"))

print("Siblings of Ram are:", result)
```

```

from kanren import Relation, facts, var, run

parent = Relation()

facts(parent,
      ("charan", "sai"),
      ("charan", "ajay"),
      ("sai", "anand"))

x = var()
result = run(0, x, parent("charan", x))

print("Children of charan are:", result)
```
