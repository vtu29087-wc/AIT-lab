```
from collections import deque

graph={
    'A':['B','C'],
    'B':[],
    'C':[]
}
visited=set()
queue=deque()
start='A'
queue.append(start)
print("BFS Traversal:")
while queue:
  node=queue.popleft()
  if node not in visited:
    visited.add(node)
    print(node,end=" ")
    for neighbour in graph.get(node,[]):
      if neighbour not in visited:
        queue.append(neighbour)
```
