```

import heapq


def h(a, b):
  return abs(a[0] - b[0]) + abs(a[1] - b[1])

def astar(grid, start, goal):
  q = [(0, start)]
  g_score = {start: 0}
  came_from = {}

  while q:
    f_score, current = heapq.heappop(q)

    if current == goal:
      path = []
      while current in came_from:
        path.append(current)
        current = came_from[current]
      path.append(start)
      return path[::-1], g_score[goal]

    for dx, dy in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
      neighbor = (current[0] + dx, current[1] + dy)

      if not (0 <= neighbor[0] < len(grid) and 0 <= neighbor[1] < len(grid[0]) and grid[neighbor[0]][neighbor[1]] == 0):
        continue

      tentative_g_score = g_score[current] + 1

      if neighbor not in g_score or tentative_g_score < g_score[neighbor]:
        came_from[neighbor] = current
        g_score[neighbor] = tentative_g_score
        f_score = tentative_g_score + h(neighbor, goal)
        heapq.heappush(q, (f_score, neighbor))

  return None, None



grid = [
    [0, 0, 0],
    [0, 1, 0],
    [0, 0, 0]
]
start = (0, 0)
end   = (2, 2)

print("Input grid:")
for row in grid:
  print(row)
print("Start:", start)
print("Goal:", end)

path, cost = astar(grid, start, end)

if path:
  print("Optimal path:", path)
  print("Cost:", cost)
else:
  print("No path found.")
```
