```
def is_safe(vertex, color, graph, colors):
  for i in range(len(graph)):
    if graph[vertex][i] == 1 and colors[i] == color:
      return False
  return True
def graph_coloring(vertex, m, graph, colors):
  if vertex == len(graph):
    return True
  for color in range(1, m + 1):
    if is_safe(vertex, color, graph, colors):
      colors[vertex] = color
      if graph_coloring(vertex + 1, m, graph, colors):
        return True
      colors[vertex] = 0
  return False
graph = [
    [0,1,1,0,0], #A
    [1,0,0,1,0], #B
    [1,0,0,1,0], #C
    [0,1,1,0,1], #D
    [0,0,0,1,0], #E
]
n = len(graph)
# The variable 'x' is not defined. Assuming it should be 'n'
x = n # Added to define 'x'

for m in range(1, x + 1):
  colors = [0] * n
  if graph_coloring(0, m, graph, colors):
    print("minimum number of colors:", m)
    print("color asssignment:")
    for i in range(n):
      print(chr(65 + i),"-> color",colors[i])
    break
```
