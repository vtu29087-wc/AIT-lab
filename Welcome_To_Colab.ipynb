```
grid = [
    ['00', '01', '02'],
    ['10', '11', '12'],
    ['20', '21', '22']
]

visited = [[False] * 3 for _ in range(3)]

# Right, Down, Left, Up
directions = [
    (0, 1),
    (1, 0),
    (0, -1),
    (-1, 0)
]

def find_paths(row, col, path):

    # Destination is 12
    if row == 1 and col == 2:
        print(" -> ".join(path))
        return

    visited[row][col] = True

    for dr, dc in directions:

        new_row = row + dr
        new_col = col + dc

        if (0 <= new_row < 3 and
            0 <= new_col < 3 and
            not visited[new_row][new_col]):

            find_paths(
                new_row,
                new_col,
                path + [grid[new_row][new_col]]
            )

    # Backtracking
    visited[row][col] = False


# Start from 00
find_paths(0, 0, ['00'])
```
